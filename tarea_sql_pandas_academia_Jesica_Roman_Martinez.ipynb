## 3. Cargar tablas a SQLite

### Punto 6. JOIN entre tablas

Muestra:
- nombre del estudiante
- nombre del curso
- estado de la inscripción

Usa `estudiantes`, `inscripciones` y `cursos`.

**Entregable:**
- SQL
- pandas

```python
# Punto 6 - SQL
# Escribe aquí tu consulta
query_6 = """
SELECT
    e.nombre,
    c.nombre_curso,
    i.estado
FROM inscripciones i
INNER JOIN estudiantes e ON i.estudiante_id = e.estudiante_id
INNER JOIN cursos c ON i.curso_id = c.curso_id
"""

sql_6 = run_sql(query_6)
sql_6
```

```python
# Punto 6 - pandas
# Escribe aquí tu solución
estudiantes_cursos = inscripciones.merge(estudiantes, on='estudiante_id')
estudiantes_cursos = estudiantes_cursos.merge(cursos, on='curso_id')
estudiantes_cursos[['nombre', 'nombre_curso', 'estado']]
```

### Punto 7. LEFT JOIN

Trae todos los estudiantes, incluso si no tuvieran inscripciones, mostrando:
- `estudiante_id`
- `nombre`
- `inscripcion_id`

**Entregable:**
- SQL
- pandas

> Pista: en este dataset puede que todos tengan al menos una inscripción, pero igual debes practicar el `LEFT JOIN`.

```python
# Punto 7 - SQL
# Escribe aquí tu consulta

query_7 = """
SELECT
    e.estudiante_id,
    e.nombre,
    i.inscripcion_id
FROM estudiantes e
LEFT JOIN inscripciones i ON e.estudiante_id = i.estudiante_id
"""
sql_7 = run_sql(query_7)
sql_7
```

```python
# Punto 7 - pandas
# Escribe aquí tu solución

resultado = estudiantes.merge(inscripciones[['estudiante_id', 'inscripcion_id']], on='estudiante_id', how='left')
resultado = resultado[['estudiante_id', 'nombre', 'inscripcion_id']]
display(resultado)
```

### Punto 8. HAVING

Calcula cuántas inscripciones tiene cada estudiante y muestra solo aquellos con **2 o más**.

Debes mostrar:
- `estudiante_id`
- `nombre`
- `n_inscripciones`

**Entregable:**
- SQL
- pandas

```python
# Punto 8 - SQL
# Escribe aquí tu consulta

query_8 = """
SELECT
    e.estudiante_id,
    e.nombre,
    COUNT(i.inscripcion_id) AS n_inscripciones
FROM estudiantes e
INNER JOIN inscripciones i ON e.estudiante_id = i.estudiante_id
GROUP BY
    e.estudiante_id,
    e.nombre
HAVING COUNT(i.inscripcion_id) >= 2
"""

sql_8 = run_sql(query_8)
sql_8
```

```python
# Punto 8 - pandas
# Escribe aquí tu solución

resultado = estudiantes.merge(inscripciones, on='estudiante_id').groupby(['estudiante_id', 'nombre'], as_index=False).agg(n_inscripciones=('inscripcion_id', 'count'))
resultado = resultado[resultado['n_inscripciones'] >= 2]

display(resultado)
```

### Punto 9. CASE WHEN / segmentación

Clasifica los cursos según su precio:
- `Bajo` si el precio es menor a 120
- `Medio` si el precio está entre 120 y 180 inclusive
- `Alto` si el precio es mayor a 180

Debes mostrar:
- `nombre_curso`
- `precio`
- `segmento_precio`

**Entregable:**
- SQL
- pandas

```python
# Punto 9 - SQL
# Escribe aquí tu consulta

query_9 = """
SELECT
    nombre_curso,
    precio,
    CASE
        WHEN precio < 120 THEN 'Bajo'
        WHEN precio BETWEEN 120 AND 180 THEN 'Medio'
        ELSE 'Alto'
    END AS segmento_precio
FROM cursos
"""

sql_9 = run_sql(query_9)
sql_9
```

```python
# Punto 9 - pandas
# Escribe aquí tu solución

resultado = cursos[['nombre_curso', 'precio']].copy()

resultado['segmento_precio'] = pd.cut(
    resultado['precio'],
    bins=[0, 120, 180, float('inf')],
    labels=['Bajo', 'Medio', 'Alto'],
    right=False
)

display(resultado)
```

### Punto 10. Ejercicio final integrador

Calcula el total pagado por categoría de curso.

Debes:
- unir `cursos`, `inscripciones` y `pagos`
- agrupar por `categoria`
- sumar `monto`
- ordenar de mayor a menor

**Entregable:**
- SQL
- pandas

```python
# Punto 10 - SQL
# Escribe aquí tu consulta

query_10 = """
SELECT
    c.categoria,
    SUM(p.monto) AS total_pagado
FROM cursos c
INNER JOIN inscripciones i ON c.curso_id = i.curso_id
INNER JOIN pagos p ON i.inscripcion_id = p.inscripcion_id
GROUP BY c.categoria
ORDER BY total_pagado DESC
"""

sql_10 = run_sql(query_10)
sql_10
```

```python
# Punto 10 - pandas
# Escribe aquí tu solución

resultado = cursos.merge(inscripciones, on='curso_id').merge(pagos, on='inscripcion_id')
resultado = resultado.groupby('categoria', as_index=False).agg(total_pagado=('monto', 'sum'))
resultado = resultado.sort_values('total_pagado', ascending=False)

display(resultado)
```
